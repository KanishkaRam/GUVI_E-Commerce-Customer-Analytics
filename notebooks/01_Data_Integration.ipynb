# Step 1 - Import Libraries

```python
import pandas as pd
import numpy as np

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

print("Libraries Imported Successfully")
```

# Step 2 - Data Integration 

```python
# Customers
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")

# Orders
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")

# Order Items
order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")

# Payments
payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")

# Reviews
reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")

# Products
products = pd.read_csv("../data/raw/olist_products_dataset.csv")

# Sellers
sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")

# Geolocation
geolocation = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")

# Category Translation
category_translation = pd.read_csv( "../data/raw/product_category_name_translation.csv")

print("All datasets loaded successfully!")
```

# Step 3 - Overview of Dataset

```python
datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews,
    "Products": products,
    "Sellers": sellers,
    "Geolocation": geolocation,
    "Category Translation": category_translation
}

for name, df in datasets.items():
    print("="*10 +' '+ name+ ' ' + "="*10 )
    print("Shape :", df.shape)
    print(df.info())
    print()
    print(df.head(3))
    print()
```

# Step 4 - Merge the Datasets

```python
# Merge 1: Customers + Orders
customer_orders = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="left"
)

print(customer_orders.shape)
customer_orders.head()
```

```python
# Merge 2: Add Order Items

customer_orders = pd.merge(
    customer_orders,
    order_items,
    on="order_id",
    how="left"
)
print(customer_orders.shape)
```

```python
# Merge 3: Add Product Details

customer_orders = pd.merge(
    customer_orders,
    products,
    on="product_id",
    how="left"
)
print(customer_orders.shape)
```

```python
# Merge 4: Add Translate Product Categories

customer_orders = pd.merge(
    customer_orders,
    category_translation,
    on="product_category_name",
    how="left"
)
print(customer_orders.shape)
```

```python
# Merge 5: add Seller Information

customer_orders = pd.merge(
    customer_orders,
    sellers,
    on="seller_id",
    how="left"
)
print(customer_orders.shape)
```

```python
# Merge 6:Add Payment Information

customer_orders = pd.merge(
    customer_orders,
    payments,
    on="order_id",
    how="left"
)
print(customer_orders.shape)
```

```python
# Merge 7: Add Review Information

customer_orders = pd.merge(
    customer_orders,
    reviews,
    on="order_id",
    how="left"
)
print(customer_orders.shape)
```

# Step 5 - Check the Final Dataset

```python
print(customer_orders.shape)

print(customer_orders.info())
customer_orders.describe(include="all").T
```

# Step 6 -  Check Missing Values

```python
missing = (customer_orders.isnull().sum().sort_values(ascending=False))
missing = missing[missing > 0]
missing
```

# Step 7 - Save the Integrated Dataset

```python
customer_orders.to_csv("../data/processed/customer_analytics_dataset.csv",index=False)
print("Integrated Dataset Saved Successfully!")
```
